```python
import pandas as pd
import openpyxl
from pathlib import Path
```

```python
'''
Definição dos caminhos do sistema para importar e exportar arquivos
Definição de um dataframe para cada tabela excel
'''

BASE_DIR = Path.cwd().parent
OUT_DIR = Path.cwd().parent / "historico_nbb" / "data"

df_26 = pd.read_excel(BASE_DIR / "historico_nbb" / "historico_seasons" / "Tabela_de_jogos_26.xlsx")
df_25 = pd.read_excel(BASE_DIR / "historico_nbb" / "historico_seasons" / "Tabela_de_jogos_25.xlsx")
df_24 = pd.read_excel(BASE_DIR / "historico_nbb" / "historico_seasons" / "Tabela_de_jogos_24.xlsx")
df_23 = pd.read_excel(BASE_DIR / "historico_nbb" / "historico_seasons" / "Tabela_de_jogos_23.xlsx")
df_22 = pd.read_excel(BASE_DIR / "historico_nbb" / "historico_seasons" / "Tabela_de_jogos_22.xlsx")
df_21 = pd.read_excel(BASE_DIR / "historico_nbb" / "historico_seasons" / "Tabela_de_jogos_21.xlsx")
df_20 = pd.read_excel(BASE_DIR / "historico_nbb" / "historico_seasons" / "Tabela_de_jogos_20.xlsx")
df_18 = pd.read_excel(BASE_DIR / "historico_nbb" / "historico_seasons" / "Tabela_de_jogos_18.xlsx")
df_17 = pd.read_excel(BASE_DIR / "historico_nbb" / "historico_seasons" / "Tabela_de_jogos_17.xlsx")
df_16 = pd.read_excel(BASE_DIR / "historico_nbb" / "historico_seasons" / "Tabela_de_jogos_16.xlsx")
```

```python
# Criação de uma função que seja capaz de fazer a limpeza para todos os dataframes
def limpeza_colunas(
        DF: pd.DataFrame,
        SEASON: int,
) -> pd.DataFrame:
    df = DF
    lista_cols = ['#', 'Season', 'Rodada', 'Casa', 'Placar_casa', 'Placar_visitante', 'Visitante', 'N1', 'Fase', 'N2', 'N3']
    df.columns = lista_cols
    mask = ['Season', 'Casa', 'Placar_casa', 'Placar_visitante', 'Visitante']
    df = df[mask]
    df = df[df['Placar_casa'] != "PLACAR"]
    df['Season'] = SEASON
    df[['Placar_casa', 'Placar_visitante']] = df['Placar_visitante'].str.extract(r'(\d+)\D+(\d+)')

    return df

df_26 = limpeza_colunas(df_26, 2026)
df_25 = limpeza_colunas(df_25, 2025)
df_24 = limpeza_colunas(df_24, 2024)
df_23 = limpeza_colunas(df_23, 2023)
df_22 = limpeza_colunas(df_22, 2022)
df_21 = limpeza_colunas(df_21, 2021)
df_20 = limpeza_colunas(df_20, 2020)
df_18 = limpeza_colunas(df_18, 2018)
df_17 = limpeza_colunas(df_17, 2017)
df_16 = limpeza_colunas(df_16, 2016)
```
